```python
import jax.numpy as jnp
import matplotlib.pyplot as plt
from jax.nn import softmax

from ising_deduplication.fitting import fit_with_adam, fit_with_sgd
from ising_deduplication.sampler import run_ising_chain
from ising_deduplication.optimize import optimize_with_adam, optimize_with_sgd
```

```python
target = jnp.array([[0.9, 0.78, 0.27], [0.78, 0.8, 0.2], [0.27, 0.2, 0.3]])
fit_carry_final, _ = fit_with_adam(target, method="surrogate")
M_est = fit_carry_final[0]
```

```python
samps = run_ising_chain(M_est, n_steps=10_000)[0]
emperical_01 = ((samps[:, 0] == 1) & (samps[:, 1] == 1)).mean()
target_01 = target[0, 1]

print(f"Empirical 01: {emperical_01:.1%}, Target 01: {target_01:.1%}")
```
